```python
#task2

import re 

def search(text):
    pattern = "\d{2}/\d{2}/\d{4}"
    if re.search(pattern,text):
        return True
    else:
        return False

print(search("today 03/08/2026"))
print(search("today 3 augast "))
            
    
```

```python
#task3
import re 
text = "jayam get 55 rupi , mansi get 200 rupi , fenil get 100 rupi , tisha get 555 rupi "

list = re.findall("\d+",text)
list
sum = 0
for i in list:
    i = int(i)
    sum+=i

print(sum)    
```
