```python
#task1

class Playlist:
    def __init__(self):
        self.__song = []

    def add_songs(self,song):
        self.__song.append(song)

    def display(self):
        print(self.__song)      


playlist = Playlist() 
playlist.add_songs("perfect")        
playlist.add_songs("tum ho")        
playlist.add_songs("rabta") 
playlist.display()       
```

```python
#task2

class Product:
    def __init__(self,name,price):
        self.name = name
        self.__price = price

    def get_price(self):
        return self.__price , self.name  

    def set_price(self,new_price):
        self.__price = new_price


product1 = Product("i phone 16",60000) 

print("get old prize and name:-",product1.get_price())

product1.set_price(65000)

print("new price :- ",product1.get_price())
```

```python
# task 5

from abc import ABC , abstractmethod

class Paymentmathod:
    @abstractmethod
    def pay(self, amount):
        pass

class paytm(Paymentmathod):
    def pay(self,amount):
        print("amount is :-",amount)

class phonepay(Paymentmathod):
    def pay(self,amount):
        print("amount is :-",amount) 

paytm = paytm()
phonepay = phonepay()

paytm.pay(6000)
phonepay.pay(5000)
```
