```python
1. .gitignore

First understand the main purpose:

.gitignore tells Git which files/folders it should NOT track.

It is mainly for files that do not belong in the repository.

For our project, .gitignore could contain:

# Python
__pycache__/
*.pyc


# Virtual environment
.venv/
venv/


# Environment/secrets
.env


# Jupyter
.ipynb_checkpoints/


# Raw data
data/raw/


# Temporary files
*.tmp
Why ignore data/raw/?

Because our raw dataset is 180 MB and we don't necessarily need to publish the entire raw dataset in the portfolio repository.

Instead, we can document:

"Download the dataset from [source] and place it in data/raw/."

So the repository might look like:

food-demand-project/
│
├── data/
│   └── raw/              ← ignored
│
├── notebooks/
│   └── EDA.ipynb
│
├── src/
│   └── feature_engineering.py
│
├── models/
├── reports/
├── README.md
└── .gitignore
2. Git LFS

Now understand the difference.

Git LFS = Git Large File Storage.

Use it when:

You have a large file that you actually WANT to keep version-controlled in the repository.

For example, suppose your trained model is:

models/demand_model.pkl

and it is 300 MB.

You might decide:

"This model is important for reproducing my project, so I want it in the repository."

Instead of storing the large binary directly in normal Git, you can use Git LFS.

Example:

git lfs install
git lfs track "*.pkl"
git add .gitattributes
git add models/demand_model.pkl
git commit -m "Add trained model using Git LFS"

Git LFS stores the large file separately while Git keeps a lightweight pointer/reference.

3. Which files go where?

This is the important part of the question.

.gitignore

Files that you don't want to track:

data/raw/*.csv
.venv/
__pycache__/
.env
.ipynb_checkpoints/
temporary files
Git LFS

Large files that you do want to track:

large trained model
large binary artifacts
large files required for reproducibility

For example:

models/demand_model.pkl

could be tracked with LFS if it is genuinely needed.
```
