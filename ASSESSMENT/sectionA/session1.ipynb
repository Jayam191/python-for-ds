Stage 1 — Problem Definition

Before touching the data, we first need to clearly understand what problem we are trying to solve.

Our scenario

We are a food delivery company and want to predict:

Which restaurant zones will face peak demand between 7 PM and 9 PM?

So we need to define exactly what "peak demand" means.

For example, suppose we decide:

A zone is considered a peak-demand zone if it has more than 500 orders between 7 PM and 9 PM.

Now our model can learn from the past two years of order data and predict which zones are likely to cross that level.

What do we decide in Problem Definition?

We define:

Goal: Predict peak-demand restaurant zones.
Time: 7 PM–9 PM.
Unit: Restaurant zone.
Historical information: Two years of order data.
Success criteria: The prediction should be accurate enough to help the company prepare delivery partners and restaurant capacity.
📦 Deliverable

The concrete deliverable from this stage is:

A clearly defined problem statement and success criteria.

For example:

"Build a model that predicts which restaurant zones will experience peak demand between 7 PM and 9 PM using historical order data."

Stage 2 — Data Collection

Now that we know what we want to predict, we need to collect the data required to make that prediction.

Our scenario

We have 2 years of order data sitting in spreadsheets.

We need to collect the information that could help us predict which zones will have high demand between 7 PM and 9 PM.

For example, our data might contain:

Order ID	Date	Time	Zone	Restaurant	Order Value
1001	10-08-2025	19:15	Zone A	Restaurant X	₹450
1002	10-08-2025	19:30	Zone B	Restaurant Y	₹300
1003	10-08-2025	20:10	Zone A	Restaurant Z	₹550

We may also collect additional useful information such as:

Day of the week
Weekend/weekday
Holidays
Promotions or discounts
Weather
Number of restaurants in each zone
Previous order volume
Why collect these?

Because demand may depend on these factors.

For example:

Friday + 7–9 PM + Promotion
             ↓
       More orders
             ↓
      Higher demand
What do we actually do?

We gather the relevant data from:

Existing spreadsheets
Order databases
Restaurant information
Promotion records
Weather data, if available

Then we combine the required information into one dataset.

📦 Deliverable

The concrete deliverable is:

A raw, consolidated dataset containing the relevant historical data.

In our scenario:

Two years of consolidated food-order data with zone, time, date, order information, and other relevant factors.

Now we have our raw data from the previous stage.

We first need to clean and transform it.

We can see problems:

We perform tasks such as:

1. Remove duplicates

Stage 4 — Exploration (EDA)

Now our data is clean and ready.

The next question is:

"What patterns and relationships exist in our data?"

This stage is called Exploratory Data Analysis (EDA).

We don't build the prediction model yet. We first understand the data.

In our food-delivery scenario

We might investigate:

1. Which zones have the highest demand?

Zone A → 400 orders
Zone B → 650 orders
Zone C → 300 orders
Zone D → 700 orders

We can see that Zone D and Zone B have higher demand.

2. Does demand increase between 7–9 PM?

We can compare order counts across different times.

5–7 PM  → 300 orders
7–9 PM  → 800 orders
9–11 PM → 450 orders

This tells us 7–9 PM may indeed be a peak period.

3. Does the day of the week matter?

For example:

Monday    → 400
Tuesday   → 420
Friday    → 650
Saturday  → 800
Sunday    → 750

We might discover that weekends have much higher demand.

4. Do promotions affect demand?

We can compare:

Promotion → 900 orders
No promo  → 500 orders

This could indicate that promotions are associated with higher demand.

What tools can we use?

We can use:

Summary statistics
Grouping and aggregation
Charts/visualizations
Correlation analysis
Distributions

The goal is to discover useful patterns before modelling.

📦 Deliverable

The concrete deliverable is:

An EDA report containing visualizations, statistics, patterns, and important insights.

For example:

"Zone D has consistently high demand during 7–9 PM, and weekend demand is significantly higher than weekday demand."

Stage 6 — Evaluation

Now we have built our prediction model.

But we can't simply say:

"The model works because it gave us a prediction."

We need to check how accurate and reliable the model actually is.

That's the purpose of Evaluation.

What do we do?

We use the unseen test data that we kept aside during modelling.

For example, suppose the actual result was:

Zone A → Not Peak
Zone B → Peak
Zone C → Not Peak
Zone D → Peak

Our model predicted:

Zone A → Not Peak ✅
Zone B → Peak ✅
Zone C → Peak ❌
Zone D → Peak ✅

We compare the prediction with the actual result.

What metrics can we use?

It depends on what we're predicting.

If we're predicting:

Peak / Not Peak

we can use classification metrics such as:

Accuracy
Precision
Recall
F1-score

If we're predicting the actual number of orders, we could use:

MAE
RMSE

The important thing is to choose metrics that match the business problem.

We also ask:

1. Is the model accurate enough?

For example:

Does it correctly identify high-demand zones?

2. Does it perform well on unseen data?

We don't want a model that only memorized the two years of historical data.

3. Is it useful for the business?

For example:

If the model predicts Zone B will be very busy, can the company actually use that information to allocate more delivery partners?

📦 Deliverable

The concrete deliverable is:

A model evaluation report and the selected final model.

The report might say:

Model performance:
Accuracy  = 89%
Precision = 87%
Recall    = 91%


Decision:
Model is suitable for peak-demand prediction.

GenAI — Stage 1: Data Preparation
Where can GenAI help?

We have two years of order data in spreadsheets. Cleaning this data can take a lot of time.

GenAI can help us generate Python/SQL code for tasks such as:

Finding missing values
Finding duplicate orders
Standardizing zone names
Converting date/time formats
Creating useful features

For example, we could ask an AI tool:

"Generate Python code to find duplicate order IDs and missing values in this dataset."

The AI can quickly generate the code.

How does this accelerate the work?

Instead of writing all the cleaning code manually, the analyst can use AI to generate a first version and then review and modify it.

⚠️ Risk

AI-generated code can be incorrect.

For example, AI might suggest deleting all rows containing missing values, even when those rows contain valuable information.

If we blindly accept the AI output:

Wrong cleaning
      ↓
Bad dataset
      ↓
Bad model
      ↓
Wrong predictions

So:

Human review is necessary before using AI-generated cleaning code.

GenAI — Stage 2: Exploration

Now we have clean data.

GenAI can help us analyze and visualize the data faster.

For example, we can ask AI to generate code for:

Demand by restaurant zone
Demand by day of week
Demand between 7–9 PM
Weekend vs weekday demand
Demand trends over time
Charts and graphs

We might ask:

"Generate Python code to visualize average 7–9 PM orders for each restaurant zone."

AI can quickly generate the visualization code.

How does this accelerate the work?

The analyst doesn't have to manually write every visualization or statistical analysis from scratch.

AI can also help summarize patterns.

For example:

"Zone B has consistently higher 7–9 PM demand on weekends."

The analyst can then investigate whether this finding is actually correct.

⚠️ Risk

The AI might misinterpret a pattern.

For example, it could say:

"Promotions cause higher demand."

But the data might only show that promotions and high demand occur together.

That's correlation, not necessarily causation.

So:

Human review is required to verify AI-generated insights before making business decisions.
