How would you use re?

You need to explain that you would create a regular expression pattern describing what a valid phone number should look like.

Then use Python's re module to find numbers matching that pattern.

For example, if our rule is:

Indian mobile number = exactly 10 digits starting with 6, 7, 8, or 9

we could use:

r'\b[6-9]\d{9}\b'

Don't worry about memorizing the symbols yet. We'll understand them.

What does this pattern mean?

r'\b[6-9]\d{9}\b'

Break it down:

\b
 ↓
start at a word boundary

[6-9]
 ↓
first digit must be 6, 7, 8, or 9

\d{9}
 ↓
then exactly 9 more digits

\b
 ↓
end at a word boundary

So the general structure is:

Start at a number boundary → require a valid first digit → require the remaining required number of digits → end at a number boundary.

Therefore:

9876543210 ✅
8123456789 ✅
5123456789 ❌
123456789 ❌
98765432101 ❌
98ABC43210 ❌

Why findall() instead of search()?

This is an important part.

Suppose our text contains:

text = """
Customer 1: 9876543210
Customer 2: 8765432109
Customer 3: 9123456780
"""

We want all valid phone numbers.

So:

re.findall(pattern, text)

returns:

['9876543210', '8765432109', '9123456780']
findall()

Means:

Find all matching values in the text.

search()
re.search(pattern, text)

means:

Find the first match in the text.

So if you need to clean a file containing many phone numbers, findall() is more appropriate.

Think:

search()
   ↓
first match

findall()
   ↓
all matches


One real-world edge case

This is the last part of the question.

A naive pattern might only accept:

9876543210

and reject:

+91 9876543210

But +91 9876543210 can be a perfectly valid Indian phone-number representation.

So a simple pattern that expects exactly 10 digits and nothing else could incorrectly reject a valid formatted number.

Another example is:

98765 43210

A pattern expecting only continuous digits might reject it even though the space is just a formatting difference.

```python
import re

text = """
Customer 1: 9876543210
Customer 2: 8765432109
Customer 3: 9123456780
Customer 4: 12345
Customer 5: 98abc43210
Customer 6: 5123456789
"""

pattern = r'\b[6-9]\d{9}\b'

valid_numbers = re.findall(pattern, text)

print("Valid phone numbers:")
print(valid_numbers)
```
