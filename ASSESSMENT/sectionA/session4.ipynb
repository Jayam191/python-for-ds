Question:
Explain how a base class with inheritance and method overriding cleanly models this.
What specific runtime problem occurs if someone writes a loop that iterates over a mixed list of
all user objects and calls the display method on each one — when that method is only defined
on some subclasses and not on the base class? Then explain how you would redesign the base
class so the loop never crashes, regardless of which new subclass is added in future

Step 1: Understand the class structure

We have 3 types of users:

User
├── Customer
├── DeliveryPartner
└── Admin

User will be the base class.

The other three will inherit from User.

Step 2: Base class + inheritance
class User:
    pass


class Customer(User):
    pass


class DeliveryPartner(User):
    pass


class Admin(User):
    pass

Here:

class Customer(User):

means Customer inherits from User.

The same applies to DeliveryPartner and Admin.

Step 3: Add display_profile()

Each user needs a different profile summary.

So each subclass can define its own version:

class User:
    pass


class Customer(User):

    def display_profile(self):
        print("Customer profile")


class DeliveryPartner(User):

    def display_profile(self):
        print("Delivery Partner profile")


class Admin(User):

    def display_profile(self):
        print("Admin profile")

Each subclass has the same method name but different behavior.

This is called method overriding.

Step 4: Use a mixed list

We can put different user objects into one list:

users = [
    Customer(),
    DeliveryPartner(),
    Admin()
]

Then:

for user in users:
    user.display_profile()

Output:

Customer profile
Delivery Partner profile
Admin profile

This is polymorphism.

The same:

user.display_profile()

does different things depending on the object.

Step 5: What is the problem?

The question specifically asks what happens if display_profile() is not present in the base class.

For example:

class User:
    pass


class Customer(User):

    def display_profile(self):
        print("Customer profile")

Now suppose we have:

users = [
    Customer(),
    User()
]

And:

for user in users:
    user.display_profile()

For Customer():

display_profile() → exists ✅

But for User():

display_profile() → doesn't exist ❌

Python raises:

AttributeError: 'User' object has no attribute 'display_profile'
Important

The problem is not simply because the list is mixed.

A mixed list is perfectly fine.

The problem is that some object in the list doesn't have the method being called.

Step 6: Redesign the base class

We can solve this by putting display_profile() in the base class:

class User:

    def display_profile(self):
        print("General user profile")

Then subclasses override it:

class Customer(User):

    def display_profile(self):
        print("Customer profile")


class DeliveryPartner(User):

    def display_profile(self):
        print("Delivery Partner profile")


class Admin(User):

    def display_profile(self):
        print("Admin profile")

Now every User has a display_profile() method.

Step 7: What if a new subclass is added?

Suppose later someone creates:

class RestaurantOwner(User):
    pass

They forgot to create their own display_profile().

That's okay.

users = [
    Customer(),
    DeliveryPartner(),
    Admin(),
    RestaurantOwner()
]

for user in users:
    user.display_profile()

Output:

Customer profile
Delivery Partner profile
Admin profile
General user profile

```python
class User:

    def display_profile(self):
        print("General user profile")


class Customer(User):

    def display_profile(self):
        print("Customer profile")


class DeliveryPartner(User):

    def display_profile(self):
        print("Delivery Partner profile")


class Admin(User):

    def display_profile(self):
        print("Admin profile")


class RestaurantOwner(User):
    pass


users = [
    Customer(),
    DeliveryPartner(),
    Admin(),
    RestaurantOwner()
]


for user in users:
    user.display_profile()
```
