```python
#task 4

class Ticket:

    def __init__(self, movie_name, price):
        self.movie_name = movie_name
        self.price = price

    def get_final_price(self):
        return self.price


class PremiumTicket(Ticket):

    def get_final_price(self):
        return super().get_final_price() + 50


ticket1 = Ticket("Avengers", 200)

ticket2 = PremiumTicket("Avengers", 200)


print("Normal Ticket Price:", ticket1.get_final_price())
print("Premium Ticket Price:", ticket2.get_final_price())
```
