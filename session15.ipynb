```python
#task1

def get_song_duration_per_minute(total_time,total_songs):
    try:
        avarage = total_time/total_songs
        print(f"avarage of the song is :- {avarage}")
    except ZeroDivisionError:
        print("error acurs on code ")

    finally:
        print("code ended here")  

get_song_duration_per_minute(200,0)              
```

```python
#task3

class NoOffersApplied(Exception):
    pass

try:
    total_spent = float(input("enter a number of total spent : -"))
    total_offer = int(input("enter an offer which is applaied :-"))
    if total_offer == 0:
        raise NoOffersApplied("no offer were applaied. case bake can not be calculated")

    avarage_case_bake = total_spent/total_offer

    print("case bake is the :-" , avarage_case_bake)

except NoOffersApplied as e:
    print("error:-",e)    
```

```python
#task4

def calculate_average_rating(total_rating, num_reviews):
    try:
        avarage = total_rating/num_reviews
        return avarage
    except ZeroDivisionError:
        print("error:- num of reviews not zero")

    finally:
        print("thank you for using calculater")    

calculate_average_rating(500,0)        
```
