```python
#task1

class song:
    def __init__(self,title,artist,duration):
        self.title = title
        self.artist = artist
        self.duration = duration
    def display(self):
        print("title",self.title)
        print("artist",self.artist)
        print("duration",self.duration)




        
```

```python
d = song("tum ho","mohit","3m")
d.display()
```
