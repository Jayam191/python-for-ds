```python
#task3

class Foodorder:
    def __init__(self,restaurant_name,item,total_price):
        self.restaurant_name = restaurant_name
        self.item = item
        self.total_price = total_price

order1 = Foodorder("zometo",
                   ["pizza","momos","pakodi"],
                   600)  

print("restaurant_name",order1.restaurant_name)     
print("item",order1.item)     
print("total_price",order1.total_price)     
```

```python
#task4


class Foodorder:
    def __init__(self,restaurant_name,item,total_price):
        self.restaurant_name = restaurant_name
        self.item = item
        self.total_price = total_price

    def add_item(self,item_name,item_price):
        self.item.append(item_name)
        self.total_price += item_price    

order1 = Foodorder("zometo",
                   ["pizza","momos","pakodi"],
                   600)  

order1.add_item("burger",200)
order1.add_item("kepechino",100)
print("restaurant_name",order1.restaurant_name)     
print("item",order1.item)     
print("total_price",order1.total_price)     
```
